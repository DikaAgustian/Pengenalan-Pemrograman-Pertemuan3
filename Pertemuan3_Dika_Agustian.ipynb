```python

# DAFTAR MENU
print("===== DAFTAR MENU =====".center(66))
print(f"{'Food':^33}{'Drink':^33}")
print("1. Nasi Goreng    : Rp35,000.00  |  1. lemongras tea : Rp13,000.00")
print("2. Chicken Katsu  : Rp37,000.00  |  2. Matcha        : Rp22,000.00")
print("3. Fried Calamari : Rp25,000.00  |  3. Americano     : Rp17,000.00")
print("4. Cookie Bomb    : Rp27,000.00  |  4. Berrycano     : Rp22,000.00")

menu = {
    "food1": 35000,
    "food2": 37000,
    "food3": 25000,
    "food4": 27000,
    "drink1": 13000,
    "drink2": 22000,
    "drink3": 17000,
    "drink4": 22000
}

nama_menu = {
    "food1": "Nasi Goreng",
    "food2": "Chicken Katsu",
    "food3": "Fried Calamari",
    "food4": "Cookie Bomb",
    "drink1": "Lemongrass Tea",
    "drink2": "Matcha",
    "drink3": "Americano",
    "drink4": "Berrycano"
}

# INPUT NAMA PELANGGAN
nama = input("\nMasukkan nama pelanggan: ")

#PENYIMPANAN DARI INPUT PESANAN PELANGGAN
keranjang = []
total = 0

# INPUT KODE BARANG
def pesan():
    while True:
        kode = input("\nMasukkan kode barang dan jumlah (contoh: drink1,3): ")

        if kode.lower() == "selesai":
            break

        data = kode.split(",")

        if len(data) != 2:
            print("Format salah! Contoh: drink1,3")
            continue

        barang = data[0].lower()

        try:
            jumlah = int(data[1])
        except ValueError:
            print("Jumlah harus berupa angka!")
            continue

        if barang in menu and jumlah > 0:
            harga = menu[barang]
            subtotal = harga * jumlah

            #PENGECEK BARANG YANG ADA DI KERANJANG
            for item in keranjang:
                if item[0] == barang:
                    item[1] += jumlah
                    item[2] += subtotal
                    break
            else:
                keranjang.append([barang, jumlah, subtotal])

            print(f"{nama_menu[barang]} x {jumlah} = Rp{subtotal:,}")
            print("Ketik selesai jika pesanan sudah lengkap")

        else:
            print("Kode barang tidak valid atau jumlah salah!")

pesan()

# HITUNG TOTAL
for barang, jumlah, subtotal in keranjang:
    total += subtotal

# SISTEM DISKON
if total >= 200000:
    diskon = 12.3
elif total >= 100000:
    diskon = 9.5
elif total >= 50000:
    diskon = 3.7
else:
    diskon = 0

potongan = total * diskon / 100
total_bayar = total - potongan


# STRUK
print()
print(f"{'=== STRUK PESANAN ===':^36}")
print(f"{nama:^36}")
print("=" * 36)

print(f"{'Nama Barang':<20}{'Jumlah':^6}{'Harga':>10}")
print("-" * 36)

for barang, jumlah, subtotal in keranjang:
    print(f"{nama_menu[barang]:<20}{jumlah:^6}{subtotal:>10,.2f}")

print("-" * 36)
print(f"{'Subtotal':<26}{total:>10,.2f}")
print(f"{'Diskon':<26}{diskon:>9}%")
print(f"{'Potongan':<26}{potongan:>10,.2f}")
print("=" * 36)
print(f"{'TOTAL BAYAR':<26}{total_bayar:>10,.2f}")
print("=" * 36)
```
